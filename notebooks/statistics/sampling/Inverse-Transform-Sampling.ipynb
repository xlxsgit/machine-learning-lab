Example: Inverse Transform Sampling from Exponential Distribution
$$ f(x) = \lambda e^{-\lambda x} \text{ for } x \geq 0 $$
$$ F(x) = 1 - e^{-\lambda x} \text{ for } x \geq 0 $$
$$ F^{-1}(u) = -\frac{1}{\lambda} \ln(1-u) $$

- step1: Generate uniform random variable U ~ Uniform(0,1)
- step2: Apply the inverse CDF transformation: X = F^{-1}(U, λ)
- step3: X follows the Exponential distribution with rate parameter λ
- step4: Visualize the sampled data against the theoretical PDF

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(42)

# Jupyter visualization settings
%config InlineBackend.figure_format = 'retina'
plt.rcParams['axes.unicode_minus'] = False
sns.set_style("whitegrid")
```

```python
def inverse_transform_sampling_exponential(lam, size):
    u = np.random.uniform(0, 1, size)
    x = - (1 / lam) * np.log(1 - u)
    return x
```

```python
# Parameters
lam = 1.0  # rate parameter
size = 10000  # number of samples
# Generate samples using inverse transform sampling
samples = inverse_transform_sampling_exponential(lam, size)
```

```python
# Visualize the results
x = np.linspace(0, 8, size)
pdf = lam * np.exp(-lam * x)
plt.figure(figsize=(8, 5))
sns.histplot(samples, bins=50, stat='density', color='skyblue', label='Sampled Data', kde=False)
plt.plot(x, pdf, 'r-', lw=2, label='Theoretical PDF')
plt.title('Inverse Transform Sampling from Exponential Distribution')
plt.xlabel('x')
plt.ylabel('Density')
plt.legend()
plt.show()
```
